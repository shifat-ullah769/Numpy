```python
# Broadcasting
'''
In NumPy describes how the library handles arrays with different shapes during arithmetic operations. Normally, operations are done element-by-element, requiring arrays to be the exact same size. Broadcasting allows the smaller array to be virtually "stretched" or expanded across the larger array so they have compatible shapes—all without making unnecessary copies of the data in memory.

Two dimensions are compatible only if:
1. They are exactly equal in size
2. One of the dimensions is exactly 1.

'''
import numpy as np
```

```python
# Scenario 1
a1 = np.arange(8).reshape(2,4)
a2 = np.arange(8,16).reshape(2,4)

print(a1)
print(a2)
```

```python
a1+a2
```

```python
# Scenario 2
a3 = np.arange(9).reshape(3,3)
a4 = np.arange(3).reshape(1,3)
print(a3,a4)
```

```python
a3+a4
```

```python
# Rules for broadcasting
'''
1. If x=m and y=n, operation will take place
2. If x=m and y=1, operation will take place (arr-1,3 and arr-4,3)
3. If x=1 and y=n, operation will take place (arr-3,1 and arr-3,4)
4. If x=1 and y=1, operation will take place (arr-1,1 and arr-4,6)
5. If x=1 and y!=n, operation will not take place (arr-1,3 and arr-4,4)
6. If x=1 and n=1 then y=m, opertion will take place (arr-1,3 and arr-3,1)
'''
```
