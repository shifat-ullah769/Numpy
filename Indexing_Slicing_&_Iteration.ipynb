```python
# Indexing, Slicing & Iteration in numpy arrays
```

```python
import numpy as np

arr1 = np.array([1,2,3,4,5])
arr2 = np.arange(24).reshape(6,4)  # reshape() --> help us to reshape our array
```

```python
# The indexing is the same in array as in lists starting from 0
arr1[-1]
arr1[3]
arr1[2:4]
```

```python
arr2[2]  # As this is a two dimensional array it will send it's 2nd index(starts from 0) element which is a one dimensional array/list/row
arr2[:4]
```

```python
# Slicing
# If we want to extract particular columns from a two dimensional array
arr2[:,2]
arr2[:,2:4]
```

```python
arr2[2:4,1:3]   # Here before the comma the indexing tells us the row index(first on included the second one excluded) and after the comma tells us the column index
```

```python
arr2[4:6,2:4]
```

```python
# Iteration
# Iterate row by row
for i in arr2:
  print(i)
```

```python
# Iterate item by item
for i in np.nditer(arr2):
  print(i)
```
