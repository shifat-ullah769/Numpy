```python
# Numpy array properties and attributes
```

```python
# Shape - It is an attribute which defines and tells the shape of an array
import numpy as np

arr1 = np.array([[1,2,3],[4,5,6]])
arr1.shape
arr2 = np.array([[[1,2],[2,3]],[[4,5],[5,6]],[[5,6],[7,8]]])
arr2
arr2.shape
```

```python
# ndim - It is an attribute which tells us the dimension of the matrix/array
arr2.ndim
arr1.ndim
```

```python
# size - This attribute tells us about the size of array or matrices means number of elements in the array/matrices
arr2.size
arr1.size
```

```python
# itemsize - This attributes tells us the memory size every elements takes to store it's value
arr2.itemsize
arr1.itemsize
```

```python
# dtype - This attribute tells us the datatype of elements in a matrices
arr1.dtype
arr2.dtype
```

```python
# astype - This attribute help us to convert the datatype of elements in a matrices/array
arr1 = arr1.astype('float64')
arr1
arr1.dtype
```
