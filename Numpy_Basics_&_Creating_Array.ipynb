```python
import numpy as np
```

```python
'''
NumPy is a Python library used for working with arrays.
It also has functions for working in domain of linear algebra, fourier transform, and matrices.

In Python we have lists that serve the purpose of arrays, but they are slow to process.
NumPy aims to provide an array object that is up to 50x faster than traditional Python lists.
The array object in NumPy is called ndarray, it provides a lot of supporting functions that make working with ndarray very easy.
Arrays are very frequently used in data science, where speed and resources are very important.

'''
```

```python
'''
There are 6 different methods by using which we can create arrays on Numpy:
1. np.array()  ---> 1D and 2D array
2. np.zeros()/ones()/empty()/random()
3. np.arrange()
4. np.linspace()
5. copy()
6. identity()

'''
```

```python
# creating numpy or n-d arrays
arr1 = np.array([1,2,3,4,5,6])
arr1
```

```python
arr2 = np.array([[1,2,3],[4,5,6]])
arr2
```

```python
arr3 = np.zeros((6,3,5))        # provide input as tuple and the input is the array/matrices size actually
arr3
```

```python
arr4 = np.ones((3,3,3))         # provide input as tuple and the input is the array/matrices size actually
arr4
```

```python
arr5 = np.identity(5)       # provide only one input which is the diagonal size means if we put 4 the matrices diagonal would be 4 one
arr5
```

```python
arr6 = np.arange(10)      # generates a list and the input is the list size. it's like a counter function of range in numpy
arr6
arr7 = np.arange(5,20)
arr7
```

```python
arr8 = np.arange(5,20,2)    # last input decides the gap between the next number here after 5 we will get 7
arr8
```

```python
arr9 = np.linspace(10,20,10)
arr9
```

```python
arr10 = arr7.copy()
arr10
```
