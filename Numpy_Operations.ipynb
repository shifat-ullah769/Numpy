```python
# Numpy operations
import numpy as np
arr1 = np.array([1,2,3,4,5,6])
arr2 = np.array([4,5,6,7,8,9])
```

```python
# Basic operations
# Substraction
arr1 - arr2
```

```python
# Vector multiplication
arr1 * arr2
```

```python
# Scaler multiplication
arr1*2
```

```python
# Conditionals
arr1>3
```

```python
# Dot product
arr3 = np.arange(6).reshape(2,3)
arr4 = np.arange(6,12).reshape(3,2)
arr3.dot(arr4)
```

```python
# Maximum and Minimum
arr4
arr4.min()
arr4.max()
arr4.min(axis=0)  #column wise min
arr4.min(axis=1)  #row wise min
```

```python
# Some basic calculations
arr4.sum()
arr4.sum(axis=0)
arr4.sum(axis=1)
arr4.mean()
arr4.std()
np.median(arr4)
np.sin(arr4)
np.exp(arr4)
```
