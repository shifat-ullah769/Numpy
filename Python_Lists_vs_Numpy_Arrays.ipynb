```python
# Less Memory - Numpy array holds less memory space rather than python lists
import sys
import numpy as np

lista = range(100)
arr1 = np.arange(100)
```

```python
print(sys.getsizeof(55)*len(lista))
```

```python
print(arr1.itemsize*arr1.size)
```

```python
# Faster - Numpy arrays are faster than python lists
import time

x = range(100000000)
y = range(100000000,200000000)

start_time = time.time()
c = [(x+y) for x,y in zip(x,y)]
print(time.time()-start_time)
```

```python
a = np.arange(100000000)
b = np.arange(100000000,200000000)

start_time = time.time()
c = a+b
print(time.time()-start_time)
```
