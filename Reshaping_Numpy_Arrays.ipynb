```python
# Reshaping numpy arrays
import numpy as np
```

```python
# Ravel - Convert higher dimension arrays to one dimension like nditer
arr1 = np.array([[2,3],[4,5],[6,7]])
arr1
arr1.ravel()
```

```python
# Reshape - Convert any one dimension array to it's factor dimension like 12 elements one dimensional array can  be converted into 3,4 or 4,3 or 2,6 etc.
arr2 = np.arange(6,24)
arr2
arr2 = arr2.reshape(2,9)
arr2
```

```python
# Transpose - Convert rows into columns and columns into rows
arr2.transpose()
```

```python
# Stacking - Combine two arrays of same size
arr3 = np.arange(1,7).reshape(2,3)
arr4 = np.arange(12,18).reshape(2,3)
np.hstack((arr3,arr4))   # horizontally combining
np.vstack((arr3,arr4))   # vertically combining
```

```python
# Splitting - Split two arrays of same size
np.hsplit(arr3,3)   # horizontal splitting means split with the column number
np.vsplit(arr3,2)   # vertical splitting means split with the row number
```

```python
# Fancy Indexing
# The exact row or column we need to extract we ca do it with this type of indexing
arr5 = np.arange(24).reshape(6,4)
arr5[[0,2,4]]
```
